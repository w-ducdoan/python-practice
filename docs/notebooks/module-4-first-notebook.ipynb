## 5. Total the loan tape's valid balances

```python
with open("loan_tape.csv") as f:
    loans = list(csv.DictReader(f))

total = 0
for loan in loans:
    value = parse_money(loan["balance"])
    if value is not None and value >= 0:
        total += value

print(f"Valid balances total ${total / 1e6:,.1f}m across {len(loans)} rows")
assert round(total / 1e6, 1) == 6198.8
```
